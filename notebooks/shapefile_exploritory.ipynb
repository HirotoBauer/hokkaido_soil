```python
from src.utils.paths import RAW_DATA_DIR, INTERM_DATA_DIR
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
shapefile_raw_dir = RAW_DATA_DIR / "shapefiles"
shapefile_proc_dir = INTERM_DATA_DIR / "shapefiles"

# Load the shapefile (.shp)
gdf = gpd.read_file(shapefile_proc_dir / "hokkaido_shape.shp")

print(gdf.columns)
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
gdf.plot(
    ax=ax,
    cmap="viridis",
    legend=True,
    edgecolor="black",
)

plt.title("Shapefile Visualization")
plt.axis("off")
plt.show()
```
